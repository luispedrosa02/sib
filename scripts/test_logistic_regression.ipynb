```python
from pathlib import Path
import sys

project = Path.cwd()
if not (project / "src" / "si").exists():
    if (project.parent / "src" / "si").exists():
        project = project.parent
    elif (project / "sib" / "src" / "si").exists():
        project = project / "sib"
    else:
        raise FileNotFoundError("Não foi possível localizar a pasta do projeto sib.")

sys.path.insert(0, str(project / "src"))

from si.io.csv_file import read_csv
from si.model_selection.split import stratified_train_test_split
from si.models.logistic_regression import LogisticRegression

print("Pasta do repositório:", project)
```

```python
dataset = read_csv(
    filename=str(project / "datasets" / "breast_bin" / "breast-bin.csv"),
    sep=",",
    features=False,
    label=True
)

print("Dimensões:", dataset.shape())
print("Classes:", dataset.get_classes())
```

```python
train, test = stratified_train_test_split(
    dataset,
    test_size=0.2,
    random_state=42
)

model = LogisticRegression(
    l2_penalty=1.0,
    alpha=0.01,
    max_iter=1000,
    patience=5,
    scale=True
)

model.fit(train)

print("Amostras de treino:", train.shape()[0])
print("Amostras de teste:", test.shape()[0])
```

```python
print(f"Accuracy no treino: {model.score(train):.4f}")
print(f"Accuracy no teste:  {model.score(test):.4f}")

print(f"Custo no treino: {model.cost(train):.4f}")
print(f"Custo no teste:  {model.cost(test):.4f}")

print("Iterações realizadas:", len(model.cost_history))
```
